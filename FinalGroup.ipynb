```python
import pandas as pd

health = pd.read_csv("health_data/healthData.csv")
education = pd.read_csv("education_data/data.csv")

# Rename rank columns
health = health.rename(columns={"Rank": "health_rank"})
education = education.rename(columns={"Rank": "education_rank"})

# Make both CSVs use the same neighborhood column name
education = education.rename(columns={"neighborhood": "neighborhood_municipality"})

# Merge on the shared neighborhood column
combined = pd.merge(health, education, on="neighborhood_municipality", how="inner")

# Add combined rank
combined["CombinedRank"] = combined["health_rank"] + combined["education_rank"]

combined.head()
```
