## Count up the data

Now we count how many arrests were made per neighborhood

```python
#Filter out all "non-major" crimes
crimrate = crime_df[crime_df.OFFENSES.str.contains('|'.join(majorcrime))]

#count the amount of rows with similar neighborhood values
offnumb = crimrate.groupby(crimrate.INCIDENTNEIGHBORHOOD.tolist(),as_index=False).size()

#Sort them values yo
offnumb=offnumb.sort_values('size', ascending=True)

#Rank number column
offnumb['Rank']=range(len(offnumb))

#Renames Columns from index and size
offnumb=offnumb.rename(columns={"index":"Neighborhoods","size":"Crime #"})

print("'Major Crime' amount per neighborhood.")
pd.set_option('display.max_rows',10)
display(offnumb)
```
